### 👀 A6 · EJEMPLO — Los dos primeros baselines (ingenuo y estacional)

Pulsa ▶ y mira lo que sale. **No hay que escribir nada.**

🧠 **Autoejercicio:** mira el resultado y responde (respuestas en la Parte B, A6).

1. ¿Qué MAE tienen el baseline ingenuo y el estacional?
2. ¿Cómo se imita el uso real en `evaluar_semanal`?
3. ¿Por qué se empieza con baselines?

> ✍️ **Respuestas del equipo:**
>
> 1. Ingenuo: **5,95** puntos. Estacional: **4,46** puntos.
> 2. En cada una de las 13 semanas de prueba, el predictor solo recibe la historia **hasta ese día** (`serie.iloc[:o]`), predice los 7 días siguientes y se compara con lo real. Es una validación *walk-forward*, igual que se usaría cada semana en producción.
> 3. Son el **listón mínimo**: simples, rápidos y fáciles de explicar. Si un modelo complejo no los supera, no merece la pena.

````python
HORIZONTE = 7                        # cada previsión cubre los 7 días siguientes (la semana que viene)
origenes = list(range(len(entrenamiento), len(serie), HORIZONTE))   # posiciones donde empieza cada semana de test (1005, 1012, ...)
origenes = [o for o in origenes if o + HORIZONTE <= len(serie)]      # descarta orígenes cuya semana se saldría del final de la serie

def mae(real, pred):   # función que mide el error absoluto medio
    # MAE = media del error absoluto |real - predicción|, en puntos de ocupación
    return float(np.mean(np.abs(np.asarray(real) - np.asarray(pred))))

def evaluar_semanal(predictor):   # simula el uso real: previsión semanal en las 13 semanas de test
    """predictor(historia: Series, origen: int) -> array de 7 valores. Devuelve (MAE, predicciones concatenadas)."""
    preds, reales = [], []                       # listas donde acumulamos predicciones y valores reales
    for o in origenes:                           # para cada semana de prueba...
        preds.append(np.asarray(predictor(serie.iloc[:o], o)))  # ...el predictor solo ve el pasado (hasta o) y predice 7 días
        reales.append(serie.iloc[o:o + HORIZONTE].values)       # ...guardamos los 7 valores reales de esa semana
    preds, reales = np.concatenate(preds), np.concatenate(reales)  # une las 13 semanas en un solo array (91 valores)
    return mae(reales, preds), preds             # devuelve el MAE total y las predicciones

def pred_ingenuo(historia, o):   # predictor 1: baseline ingenuo
    # Baseline ingenuo: repite 7 veces el último valor conocido ("mañana = hoy")
    return np.repeat(historia.iloc[-1], HORIZONTE)

def pred_estacional(historia, o):   # predictor 2: baseline estacional (t-7)
    # Baseline estacional ingenuo: copia la última semana (el lunes que viene = el último lunes)
    return historia.iloc[-7:].values

mae_ing, p_ing = evaluar_semanal(pred_ingenuo)        # evalúa el ingenuo
mae_est, p_est = evaluar_semanal(pred_estacional)     # evalúa el estacional
print(f"MAE ingenuo    : {mae_ing:.2f} puntos")       # :.2f = 2 decimales
print(f"MAE estacional : {mae_est:.2f} puntos")   # imprime el MAE del estacional con 2 decimales
````

````python
def exogenas(indice):   # función que crea las variables exógenas para unas fechas
    # Construye las variables externas (exógenas) CONOCIDAS de antemano para unas fechas dadas
    dia_anio = indice.dayofyear.values           # número de día del año (1..366)
    X = pd.DataFrame(index=indice)               # tabla vacía con las fechas como índice
    for k in (1, 2):                             # 2 armónicos de Fourier: k=1 onda anual, k=2 onda semestral
        X[f"sin{k}"] = np.sin(2 * np.pi * k * dia_anio / 365.25)   # seno del ciclo anual
        X[f"cos{k}"] = np.cos(2 * np.pi * k * dia_anio / 365.25)   # coseno del ciclo anual
    X["festivo"] = ocup.loc[indice, "festivo"].values              # 1 si es festivo, 0 si no (se sabe por calendario)
    return X   # devuelve la tabla de exógenas (sin1, cos1, sin2, cos2, festivo)

X_todo = exogenas(serie.index)          # exógenas para TODAS las fechas (entreno + prueba)
X_ent = X_todo.iloc[:-DIAS_TEST]        # solo la parte de entrenamiento

# SARIMAX: order=(p,d,q)=(1,0,1) → 1 AR, sin diferencia, 1 MA
# seasonal_order=(P,D,Q,s)=(1,1,1,7) → parte estacional semanal con 1 diferencia estacional (resta el valor de hace 7 días)
modelo = SARIMAX(entrenamiento, exog=X_ent, order=(1, 0, 1), seasonal_order=(1, 1, 1, 7),
                 enforce_stationarity=False, enforce_invertibility=False)   # no forzar restricciones: ajuste más flexible
res = modelo.fit(disp=False, maxiter=200)   # ajusta (estima coeficientes); disp=False = sin mensajes; máx. 200 iteraciones
print(res.summary().tables[1])             # tabla de coeficientes (valor, error, z, p-valor, intervalo 95%)
````

## 5. ARIMA y SARIMAX

#### A10 · Ejemplo — Ajustar un modelo SARIMAX con festivos y Fourier

**Idea sin fórmulas.**
- **AR (autorregresivo)**: el valor de hoy depende de los de días anteriores.
- **I (integrado)**: se trabaja con *diferencias* para quitar tendencias.
- **MA (media móvil)**: el valor depende de los errores recientes del modelo.
- **S (estacional)**: lo mismo pero mirando "hace 7 días".
- **X (exógenas)**: variables externas que conocemos de antemano, como los **festivos**.

Como la estacionalidad anual (invierno alto, verano bajo) es larga para SARIMAX, la representamos con
dos ondas seno/coseno del año (**términos de Fourier**), que son calendario y por tanto conocidos en el futuro.
**Cuidado:** no usamos la temperatura como variable, porque no la conoceríamos con antelación (sería *fuga*).

- **Qué hace:**
    - `exogenas(indice)` construye las variables externas conocidas de antemano: dos ondas seno/coseno del año (términos de Fourier, `sin1`, `cos1`, `sin2`, `cos2`) y los festivos.
    - `SARIMAX(entrenamiento, exog=X_ent, order=(1, 0, 1), seasonal_order=(1, 1, 1, 7), ...)` define el modelo (AR 1, MA 1 y parte estacional semanal) y `.fit(...)` lo ajusta (`res`).
    - `res.summary().tables[1]` imprime la tabla de coeficientes.
- **Qué debe salir:**
    - Una tabla de coeficientes (`sin1`, `cos1`, `sin2`, `cos2`, `festivo`, `ar.L1`, `ma.L1`, `ar.S.L7`, `ma.S.L7`, `sigma2`). Tarda unos segundos.
- **Cómo leerlo:**
    - El coeficiente de `festivo` es **+9,2**: un festivo suma unos 9 puntos de ocupación (con un valor `P>|z|` de 0,000, es decir, claramente distinto de 0). `cos1` (+8,0) recoge el patrón anual (alto en invierno). El término `ar.S.L7` es casi 0 (0,003, `P>|z|` = 0,52) y no aporta.
- **¿De dónde sale cada valor?**
    - **`order=(1, 0, 1)`** → es el punto de partida más sencillo con memoria: 1 término autorregresivo (el valor de ayer influye en hoy) y 1 de media móvil (el error de ayer). Medido con los **últimos 91 días del entrenamiento** (no con la prueba), error medio / AIC (menor es mejor): sin partes AR/MA, el ajuste falla del todo (≈ 77 puntos) · (1,0,1) sin parte estacional 4,72 / 5 855 · **(1,0,1)(1,1,1,7) 3,48 / 5 064** · (2,0,2)(1,1,1,7) 3,55 / 5 088 · (1,1,1)(1,1,1,7) 3,49 / 5 061 · (0,0,1)(1,1,1,7) 3,59 / 5 094. Subir los órdenes **no mejora**: el modelo simple ya está en el grupo de cabeza.
    - **`seasonal_order=(1, 1, 1, 7)`** → el último 7 es el **periodo**: la gráfica muestra un ciclo semanal. Sin la parte estacional el error sube de 3,48 a 4,72. Con periodo 14 sale 3,51: igual, pero tarda más del doble en ajustarse. El `1` central (diferencia estacional) resta el valor de hace 7 días para quitar el ciclo semanal.
    - **2 pares seno/coseno (Fourier)** → recogen el ciclo anual sin 365 variables. Medido: 0 armónicos → 3,64 · 1 → 3,62 · **2 → 3,48** · 3 → 3,31 · 4 → 3,27. **Más armónicos mejoran algo**; 2 es un compromiso (cada armónico añade 2 columnas y más riesgo de sobreajuste). Con 3 años de datos, subir a 3 o 4 sería razonable.
    - **`maxiter=200`** → el número máximo de iteraciones del optimizador: con 200 sobra para que converja; si faltaran aparecería un aviso de convergencia.
- **Si falla:**
    - `NameError: name 'entrenamiento' is not defined` → falta el ejemplo de la partición temporal.

2. *¿Por qué se usan términos de Fourier en vez de una estacionalidad anual de 365 días?* → Porque una estacionalidad tan larga es **pesada para SARIMAX**; dos ondas seno/coseno del año la representan con pocos parámetros, y son calendario (se conocen en el futuro).

## Resumen: qué has aprendido
- Las series se parten **por fecha** y se evalúan imitando el uso real (previsión a 7 días).
- Siempre hay que compararse con un **baseline**; a veces gana o casi.
- La descomposición separa tendencia, estacionalidad y resto; SARIMAX combina memoria, estacionalidad y variables conocidas.
- Evita usar como entrada datos que no tendrás en el momento de predecir.
